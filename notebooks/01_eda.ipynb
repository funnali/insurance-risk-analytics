```python
import pandas as pd

file_path = "../data/MachineLearningRating_v3.txt"

df = pd.read_csv(
    file_path,
    sep="|",
    low_memory=False
)

df.head()
```

```python
# Dataset shape
print("Dataset Shape:")
print(df.shape)

print("\n")

# Column names
print("Columns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
# Loss Ratio
df["LossRatio"] = df["TotalClaims"] / df["TotalPremium"]

# Margin
df["Margin"] = df["TotalPremium"] - df["TotalClaims"]

df[["LossRatio", "Margin"]].head()
```

```python
overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:")
print(round(overall_loss_ratio, 4))
```
