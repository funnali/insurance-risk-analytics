```python
import pandas as pd

file_path = "../data/MachineLearningRating_v3.txt"

df = pd.read_csv(
    file_path,
    sep="|",
    low_memory=False
)

df.head()
```

```python
# Dataset shape
print("Dataset Shape:")
print(df.shape)

print("\n")

# Column names
print("Columns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
# Loss Ratio
df["LossRatio"] = df["TotalClaims"] / df["TotalPremium"]

# Margin
df["Margin"] = df["TotalPremium"] - df["TotalClaims"]

df[["LossRatio", "Margin"]].head()
```

```python
overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:")
print(round(overall_loss_ratio, 4))
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

plt.style.use("default")
sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
```

```python
df["TransactionMonth"] = pd.to_datetime(df["TransactionMonth"])

df["TransactionMonth"].head()
```

```python
missing_values = df.isnull().sum()

missing_percent = (
    missing_values / len(df)
) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percent
})

missing_df.sort_values(
    by="Percentage",
    ascending=False
).head(15)
```

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    df["TotalClaims"],
    bins=50
)

plt.title("Distribution of Total Claims")
plt.xlabel("Total Claims")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    x=df["TotalClaims"]
)

plt.title("Boxplot of Total Claims")

plt.show()
```

```python
province_risk = (
    df.groupby("Province")[["TotalClaims", "TotalPremium"]]
    .sum()
)

province_risk["LossRatio"] = (
    province_risk["TotalClaims"] /
    province_risk["TotalPremium"]
)

province_risk = province_risk.sort_values(
    by="LossRatio",
    ascending=False
)

province_risk.head(10)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=province_risk.index,
    y=province_risk["LossRatio"]
)

plt.xticks(rotation=45)

plt.title("Loss Ratio by Province")
plt.xlabel("Province")
plt.ylabel("Loss Ratio")

plt.show()
```

```python
gender_risk = (
    df.groupby("Gender")[["TotalClaims", "TotalPremium"]]
    .sum()
)

gender_risk["LossRatio"] = (
    gender_risk["TotalClaims"] /
    gender_risk["TotalPremium"]
)

gender_risk.sort_values(
    by="LossRatio",
    ascending=False
)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=gender_risk.index,
    y=gender_risk["LossRatio"]
)

plt.title("Loss Ratio by Gender")
plt.xlabel("Gender")
plt.ylabel("Loss Ratio")

plt.show()
```

```python
vehicle_risk = (
    df.groupby("VehicleType")[["TotalClaims", "TotalPremium"]]
    .sum()
)

vehicle_risk["LossRatio"] = (
    vehicle_risk["TotalClaims"] /
    vehicle_risk["TotalPremium"]
)

vehicle_risk = vehicle_risk.sort_values(
    by="LossRatio",
    ascending=False
)

vehicle_risk.head(10)
```

```python
plt.figure(figsize=(14, 6))

sns.barplot(
    x=vehicle_risk.index,
    y=vehicle_risk["LossRatio"]
)

plt.xticks(rotation=75)

plt.title("Loss Ratio by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Loss Ratio")

plt.show()
```

```python
monthly_trends = (
    df.groupby("TransactionMonth")[["TotalClaims", "TotalPremium"]]
    .sum()
)

monthly_trends["LossRatio"] = (
    monthly_trends["TotalClaims"] /
    monthly_trends["TotalPremium"]
)

monthly_trends.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_trends.index,
    monthly_trends["TotalClaims"],
    label="Total Claims"
)

plt.plot(
    monthly_trends.index,
    monthly_trends["TotalPremium"],
    label="Total Premium"
)

plt.legend()

plt.title("Claims and Premium Trends Over Time")
plt.xlabel("Transaction Month")
plt.ylabel("Amount")

plt.show()
```

```python
numeric_cols = df.select_dtypes(
    include=["int64", "float64"]
)

correlation_matrix = numeric_cols.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Which vehicle makes have highest/lowest claim amounts?
make_risk = df.groupby('make').agg(
    AvgClaim=('TotalClaims', 'mean'),
    TotalClaims=('TotalClaims', 'sum'),
    PolicyCount=('PolicyID', 'count')
).reset_index()

make_risk = make_risk[make_risk['PolicyCount'] > 100]  # filter small samples

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Top 10 highest avg claim
top10 = make_risk.nlargest(10, 'AvgClaim')
axes[0].barh(top10['make'], top10['AvgClaim'], color='crimson')
axes[0].set_title('Top 10 Makes — Highest Avg Claim', fontweight='bold')
axes[0].set_xlabel('Average Claim (R)')

# Top 10 lowest avg claim
bot10 = make_risk[make_risk['AvgClaim'] > 0].nsmallest(10, 'AvgClaim')
axes[1].barh(bot10['make'], bot10['AvgClaim'], color='steelblue')
axes[1].set_title('Top 10 Makes — Lowest Avg Claim', fontweight='bold')
axes[1].set_xlabel('Average Claim (R)')

plt.tight_layout()
plt.savefig('../reports/vehicle_make_risk.png', dpi=150, bbox_inches='tight')
plt.show()

# Also show top/bottom 5 in text
print("=== Highest Risk Makes ===")
print(make_risk.nlargest(5, 'AvgClaim')[['make', 'AvgClaim', 'PolicyCount']])
print("\n=== Lowest Risk Makes ===")
print(make_risk[make_risk['AvgClaim'] > 0].nsmallest(5, 'AvgClaim')[['make', 'AvgClaim', 'PolicyCount']])
```
