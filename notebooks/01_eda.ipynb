```python
import pandas as pd

file_path = "../data/MachineLearningRating_v3.txt"

df = pd.read_csv(
    file_path,
    sep="|",
    low_memory=False
)

df.head()
```

```python
# Dataset shape
print("Dataset Shape:")
print(df.shape)

print("\n")

# Column names
print("Columns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
# Loss Ratio
df["LossRatio"] = df["TotalClaims"] / df["TotalPremium"]

# Margin
df["Margin"] = df["TotalPremium"] - df["TotalClaims"]

df[["LossRatio", "Margin"]].head()
```

```python
overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:")
print(round(overall_loss_ratio, 4))
```

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

plt.style.use("default")
sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
```

```python
df["TransactionMonth"] = pd.to_datetime(df["TransactionMonth"])

df["TransactionMonth"].head()
```

```python
missing_values = df.isnull().sum()

missing_percent = (
    missing_values / len(df)
) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percent
})

missing_df.sort_values(
    by="Percentage",
    ascending=False
).head(15)
```

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    df["TotalClaims"],
    bins=50
)

plt.title("Distribution of Total Claims")
plt.xlabel("Total Claims")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    x=df["TotalClaims"]
)

plt.title("Boxplot of Total Claims")

plt.show()
```

```python
province_risk = (
    df.groupby("Province")[["TotalClaims", "TotalPremium"]]
    .sum()
)

province_risk["LossRatio"] = (
    province_risk["TotalClaims"] /
    province_risk["TotalPremium"]
)

province_risk = province_risk.sort_values(
    by="LossRatio",
    ascending=False
)

province_risk.head(10)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=province_risk.index,
    y=province_risk["LossRatio"]
)

plt.xticks(rotation=45)

plt.title("Loss Ratio by Province")
plt.xlabel("Province")
plt.ylabel("Loss Ratio")

plt.show()
```
